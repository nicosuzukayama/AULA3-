#QUESTÃO 02 – Soma dos Números

```python
# Solicite ao usuário um número inteiro positivo.
# Utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

try:
    num = int(input("Digite um número inteiro positivo: "))
    if num < 1:
        print("Por favor, digite um número inteiro positivo.")
    else:
        soma = 0
        for i in range(1, num + 1):
            soma += i
        print(f"A soma de todos os números de 1 até {num} é: {soma}")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

# QUESTÃO 03 – Tabuada

```python
# Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

try:
    num = int(input("Digite um número inteiro: "))
    print(f"Tabuada do {num}:")
    for i in range(1, 11):
        produto = num * i
        print(f"{num} x {i} = {produto}")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

# QUESTÃO 04 – Fatorial

```python
# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

try:
    num = int(input("Digite um número inteiro positivo: "))

    if num < 0:
        print("Fatorial não é definido para números negativos.")
    elif num == 0:
        print("O fatorial de 0 é 1.")
    else:
        fatorial = 1
        for i in range(1, num + 1):
            fatorial *= i
        print(f"O fatorial de {num} é: {fatorial}")
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")
```

# QUESTÃO 05 – Caixa Eletrônico

```python
# Um caixa eletrônico possui notas de:R$ 100; R$ 50; R$ 20; R$ 10; R$ 5; R$ 2; R$ 1.

# Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

try:
    saque = int(input("Digite o valor do saque (apenas números inteiros): R$"))

    if saque < 0:
        print("Não é possível sacar valores negativos.")
    elif saque == 0:
        print("Nenhum valor a sacar.")
    else:
        notas = [100, 50, 20, 10, 5, 2, 1]
        quantidade_notas = {}
        valor_restante = saque

        print(f"Para o saque de R${saque}, as notas serão:")

        for nota in notas:
            if valor_restante >= nota:
                quantidade = valor_restante // nota
                quantidade_notas[nota] = quantidade
                valor_restante %= nota
                print(f"- {quantidade} nota(s) de R${nota}")

        if valor_restante > 0:
            print(f"\nSobrou R${valor_restante}. O caixa não consegue pagar este valor exato com as notas disponíveis.")

except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro para o valor do saque.")
```
