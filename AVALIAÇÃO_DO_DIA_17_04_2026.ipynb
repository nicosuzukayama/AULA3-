1- Par ou ímpar
# Peça um número ao usuário e informe se ele é par ou ímpar.

```python
# Entrada

a = int(input("Escolha um número: "))

# IDENTIFICAÇÃO

if a % 2 == 0:
    print(f"{a} é um número par")
else:
    print(f"{a} é um número ímpar")

# Saída

print("+-+-+-+-+-+-+-+-+-+-+-+-+-+-")
```

2- Maior de dois números
# Solicite dois números e mostre qual é o maior (ou se são iguais).

```python
# Solicitando os números ao usuário

# Usamos float() para permitir números com vírgulas/pontos decimais

numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

# Lógica de comparação

if numero1 > numero2:
    print(f"O maior número é: {numero1}")
elif numero2 > numero1:
    print(f"O maior número é: {numero2}")
else:
    print("Os dois números são iguais.")
```

3- Aprovação de aluno
# Peça a nota de um aluno e informe:

Aprovado (nota ≥ 7)

Recuperação (5 ≤ nota &lt; 7)

Reprovado (nota &lt; 5)

```python
# Solicitando a nota do aluno

# Usamos float() pois notas podem ter casas decimais (ex: 7.5)

nota = float(input("Digite a nota do aluno: "))

# Lógica de verificação

if nota >= 7:
    print("Situação: Aprovado")
elif 5 <= nota < 7:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
```

4- Classificação de idade
# Receba a idade e classifique:
Criança (0–12)

Adolescente (13–17)

Adulto (18–59)

Idoso (60+)

```python
# Recebendo a idade do usuário

# Usamos int() porque idade geralmente é um número inteiro

idade = int(input("Digite a idade: "))

# Lógica de classificação

if idade < 0:
    print("Idade inválida! Por favor, digite um valor positivo.")

elif idade <= 12:
    print("Classificação: Criança")
elif idade <= 17:
    print("Classificação: Adolescente")
elif idade <= 59:
    print("Classificação: Adulto")
else:
    print("Classificação: Idoso")
```

7- Calculadora simples
# Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o resultado.

```python
# Solicitando os números ao usuário

numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

# Solicitando a operação desejada

print("Operações disponíveis: +, -, *, /")
operacao = input("Digite a operação desejada: ")

# Lógica da calculadora

if operacao == "+":
    resultado = numero1 + numero2
    print(f"Resultado: {numero1} + {numero2} = {resultado}")
elif operacao == "-":
    resultado = numero1 - numero2
    print(f"Resultado: {numero1} - {numero2} = {resultado}")
elif operacao == "*":
    resultado = numero1 * numero2
    print(f"Resultado: {numero1} * {numero2} = {resultado}")
elif operacao == "/":

    # Verificação para evitar divisão por zero

    if numero2 != 0:
        resultado = numero1 / numero2
        print(f"Resultado: {numero1} / {numero2} = {resultado}")
    else:
        print("Erro: Não é possível dividir por zero!")
else:
    print("Operação inválida! Por favor, escolha entre +, -, * ou /.")
```

8- Dia da semana
# Peça um número de 1 a 7 e mostre o dia correspondente
(ex: 1 = domingo).

```python
# Solicitando um número de 1 a 7

numero = int(input("Digite um número de 1 a 7 para saber o dia da semana: "))

# Lógica de verificação

if numero == 1:
    print("1 = Domingo")
elif numero == 2:
    print("2 = Segunda-feira")
elif numero == 3:
    print("3 = Terça-feira")
elif numero == 4:
    print("4 = Quarta-feira")
elif numero == 5:
    print("5 = Quinta-feira")
elif numero == 6:
    print("6 = Sexta-feira")
elif numero == 7:
    print("7 = Sábado")
else:
    print("Valor inválido! Por favor, digite um número entre 1 e 7.")
```

9- Ano bissexto
# Solicite um ano e informe se ele é bissexto ou não.

```python
# Solicitando o ano ao usuário

ano = int(input("Digite um ano para verificar se é bissexto: "))

# Lógica de verificação do ano bissexto:

# 1. Deve ser divisível por 4

# 2. NÃO pode ser divisível por 100, A MENOS que também seja divisível por 400

if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"O ano {ano} é BISSEXTO! ")
else:
    print(f"O ano {ano} NÃO é bissexto.")
```

10- Desconto em compras
# Peça o valor de uma compra:

≥ 100: 10% de desconto

≥ 50 e &lt; 100: 5% de desconto

&lt; 50: sem desconto

```python
# Solicitando o valor total da compra

valor_compra = float(input("Digite o valor total da compra: R$ "))

# Lógica para aplicar o desconto

if valor_compra >= 100:
    desconto = 0.10 # 10%
    print("Você ganhou 10% de desconto!")
elif valor_compra >= 50:
    desconto = 0.05 # 5%
    print("Você ganhou 5% de desconto!")
else:
    desconto = 0
    print("Compras abaixo de R$ 50.00 não possuem desconto.")

# Cálculos finais

valor_desconto = valor_compra * desconto
valor_final = valor_compra - valor_desconto

# Exibindo os resultados

print(f"Valor do desconto: R$ {valor_desconto:.2f}")
print(f"Total a pagar: R$ {valor_final:.2f}")
```
