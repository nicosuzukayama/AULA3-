# Exercicíos de "for"

```python
#Escrever um programa que imprima so números de 1 a 10
for i in range (1, 11):
  print(i)
```

```python
#Escrever um programa que imprima os números de 10 a 1 contagem regressiva
for i in range (1, 11):
  print(11 -i)
```

```python
palavra = "phyton"
for i in (palavra):
  print (i)
```

```python
#Escrever um programa que imprima a palavra phyton ao contrario
palavra = "phyton"
for i in palavra[::-1]:
    print (i)
```

```python
#Escrever um programa que descubra se o númer é par ou ímpar
a = int (input("digite um número"))
resto = a % 2
if resto == 0:
  print (f"O número {a} é par")
else:
  print (f"O número {a} é ímpar")
```

# EXERCICÍOS DE 1 à 10

```python
# contagem simples, o usuario digita um número e o programa imprimi do zero até o número
a = int(input("Digite um número: "))

for i in range (0, a + 1) :
  print(i)
```

```python
# contagem regressiva, o usuario digita um número e o programa imprimi do número até zero
a = int(input("Digite um número para iniciar a contagem regressiva: "))
for i in range (a, -1, -1) :
  print(i)
```

```python
# Números pares, o usuário digita um número e o programa imprimi os números pares
a = int(input("Digite um número: "))

for i in range(a + 1):
    if i % 2 == 0:
        print(i)
```

```python
# Números impares, o usuário digita um número e o programa imprimi os números impares
a = int(input("digite um número: "))

for i in range(a + 1):
    if i % 2 != 0:
        print(i)
```

```python
# Soma, o usuário digita um número e o programa soma os números do zero até o número
numero_digitado = int(input("Digite um número: "))
soma = 0
for i in range(numero_digitado + 1):
    soma += i
print(f"A soma dos números de 0 até {numero_digitado} é: {soma}")
```

```python
# Tabuada, o usuário digita um número e o programa imprimi a tabuada do número
a = int(input("Digite um número para ver a tabuada: "))

for i in range(1, 11):
    print(f"{a} x {i} = {a * i}")
```

```python
# contar letras, o usuário digita uma palavra e o programa conta as letras da palavra
palavra = input("Digite uma palavra: ")
quantidade_letras = len(palavra)
print(f"A palavra '{palavra}' tem {quantidade_letras} letras.")
```

```python
# contar vogais, o usário digita uma palavra e o programa conta as vogais
palavra = input("Digite uma palavra: ")
vogais = "aeiouAEIOU"
contador_vogais = 0

for letra in palavra:
    if letra in vogais:
        contador_vogais += 1
print(f"A palavra '{palavra}' tem {contador_vogais} vogais.")
```

```python
# maior número digitado, o usuário digita quatro números e o programa imprimi o maior número digitado
num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))
num3 = int(input("Digite o terceiro número: "))
num4 = int(input("Digite o quarto número: "))

maior_numero = max(num1, num2, num3, num4)
print(f"O maior número digitado é: {maior_numero}")
```

```python
# o usuário digita um número entre 01 e 100 e o programa mostra quais são divisiveis por 3 e 5
numero_limite = int(input("Digite um número entre 1 e 100: "))

if 1 <= numero_limite <= 100:
    print(f"Números divisíveis por 3 e 5 até {numero_limite}:")
    for i in range(1, numero_limite + 1):
        if i % 3 == 0 and i % 5 == 0:
            print(i)
else:
    print("Por favor, digite um número entre 1 e 100.")
```
