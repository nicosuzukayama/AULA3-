Par ou ímpar
# Peça um número ao usuário e informe se ele é par ou ímpar.

```python
# Entrada

a = int(input("Escolha um número: "))

# IDENTIFICAÇÃO

if a % 2 == 0:
    print(f"{a} é um número par")
else:
    print(f"{a} é um número ímpar")

# Saída

print("+-+-+-+-+-+-+-+-+-+-+-+-+-+-")
```

Maior de dois números
# Solicite dois números e mostre qual é o maior (ou se são iguais).

```python
# Entrada

num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))

# Comparação e Saída

if num1 > num2:
    print(f"O maior número é: {num1}")
elif num2 > num1:
    print(f"O maior número é: {num2}")
else:
    print("Os números são iguais.")
```

```python
# Solicitando os números ao usuário
# Usamos float() para permitir números com vírgulas/pontos decimais
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
# Lógica de comparação
if num1 > num2:
    print(f"O maior número é: {num1}")
elif num2 > num1:
    print(f"O maior número é: {num2}")
else:
    print("Os dois números são iguais.")
```
